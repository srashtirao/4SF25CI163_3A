```python
df["delivery_review_count"] = pd.to_numeric(df["delivery_review_count"], errors="coerce")
df
```

```python
df.drop(
    columns=["restId", "url", "location_longitude", "location_latitude"],inplace=True)

```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
numeric_columns = df.select_dtypes(
    include=["number"]).columns
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].mean())
df.isnull().sum()
```

```python
categorical_columns = df.select_dtypes(
    include=["object"]).columns
for col in categorical_columns:
    df[col] = df[col].fillna("Not Available")
df.isnull().sum().sum()
```

```python
top_restaurants = df.sort_values(
by="aggregate_rating", ascending=False)
print(top_restaurants[["name", "aggregate_rating"]].head(10))
```

```python
average_rating = df["aggregate_rating"].mean()
print(average_rating)
```

```python
cuisine_data = df["cuisine"].str.split(",").explode()
top_cuisines = cuisine_data.value_counts().head(10)
print(top_cuisines)
```

```python
popular_restaurants = df.sort_values(by="rating_votes", ascending=False)
print(popular_restaurants[["name", "rating_votes"]].head(10))
```

```python
print("Average Dining Rating:",df["dining_rating"].mean())
print("Average Delivery Rating:",
df["delivery_rating"].mean())
```

```python
top_cuisines.index = top_cuisines.index.str.strip()
plt.figure(figsize=(10, 5))
top_cuisines.plot(kind="bar")
plt.title("Top 10 Popular Cuisines")
plt.xlabel("Cuisines")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["aggregate_rating"], bins=10)
plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Restaurant Ratings")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
ratings = [
    df["dining_rating"].mean(),
    df["delivery_rating"].mean()
]
labels = ["Dining Rating", "Delivery Rating"]
plt.figure(figsize=(7, 5))
plt.bar(labels, ratings)
plt.title("Average Dining vs Delivery Ratings")
plt.xlabel("Rating Type")
plt.ylabel("Average Rating")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["approx_cost"].dropna(), bins=15)
plt.title("Restaurant Cost Distribution")
plt.xlabel("Approximate Cost (Rs.)")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df["rating_votes"],
    df["aggregate_rating"]
)
plt.title("Rating Votes vs Restaurant Ratings")
plt.xlabel("Rating Votes")
plt.ylabel("Aggregate Rating")
plt.show()
```
